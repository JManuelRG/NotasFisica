# Fricción (Fuerzas de Rozamiento)

La **fricción** o **rozamiento** es una fuerza de contacto que se opone al movimiento relativo de deslizamiento entre dos superficies. Según el enfoque clásico del libro **Sears & Zemansky (Física Universitaria)**, cuando un cuerpo descansa o se desliza sobre una superficie, podemos descomponer la fuerza de contacto ejercida por la superficie sobre el cuerpo en dos componentes:

1.  **Fuerza Normal ($\vec{n}$):** Perpendicular a las superficies en contacto.
2.  **Fuerza de Fricción ($\vec{f}$):** Tangente a las superficies en contacto y paralela a la dirección del movimiento (o intento de movimiento).

---

## Tipos de Fricción

### 1. Fricción Estática ($\vec{f}_s$)
Actúa cuando los cuerpos están en reposo relativo. Esta fuerza es de magnitud variable y se ajusta para impedir que el cuerpo comience a moverse, hasta alcanzar un valor máximo admisible:

$$f_s \le f_{s,\text{máx}} = \mu_s n$$

*   Donde $\mu_s$ es el **coeficiente de fricción estática** (adimensional).
*   $n$ es la magnitud de la fuerza normal.

### 2. Fricción Cinética o Dinámica ($\vec{f}_k$)
Actúa una vez que se inicia el movimiento de deslizamiento relativo entre las superficies. Su magnitud es aproximadamente constante y es proporcional a la fuerza normal:

$$f_k = \mu_k n$$

*   Donde $\mu_k$ es el **coeficiente de fricción cinética**.
*   Por lo general, para un mismo par de superficies, se cumple que: **$\mu_k < \mu_s$** (es más fácil mantener un objeto en movimiento que iniciar su deslizamiento desde el reposo).

---

### Propiedades Clave (Sears-Zemansky):
*   **Dirección:** La dirección de la fuerza de fricción siempre es opuesta al movimiento relativo (en el caso de la fricción cinética) o a la dirección en la que el cuerpo tendería a moverse bajo la acción de otras fuerzas (en el caso de la fricción estática).
*   **Independencia del Área:** A nivel macroscópico, las fuerzas de fricción de deslizamiento son casi independientes del área de contacto aparente entre las superficies.

### Ejercicio 1: Bloque en reposo sobre superficie horizontal
Un bloque de madera de $10\text{ kg}$ reposa sobre una mesa horizontal. El coeficiente de fricción estática entre el bloque y la mesa es $\mu_s = 0.4$. Si se aplica una fuerza horizontal de $30\text{ N}$, determine si el bloque se mueve y calcule la fuerza de fricción ejercida sobre él. (Use $g = 9.8\text{ m/s}^2$).

**Solución:**
1. **Identificar fuerzas normales:** En el eje vertical, como no hay aceleración:
   $$n = mg = (10\text{ kg})(9.8\text{ m/s}^2) = 98\text{ N}$$
2. **Calcular la fuerza de fricción estática máxima ($f_{s,\text{máx}}$):**
   $$f_{s,\text{máx}} = \mu_s n = (0.4)(98\text{ N}) = 39.2\text{ N}$$
3. **Analizar el movimiento:** La fuerza aplicada ($F = 30\text{ N}$) es menor que la fuerza necesaria para romper el estado de reposo ($f_{s,\text{máx}} = 39.2\text{ N}$).
4. **Resultado:** El bloque **no se mueve**. Por lo tanto, por equilibrio de fuerzas en el eje horizontal, la fuerza de fricción estática real es exactamente igual a la fuerza aplicada:
   $$f_s = 30\text{ N}$$

### Ejercicio 2: Fuerza mínima para iniciar el movimiento
Con los datos del Ejercicio 1 (bloque de $10\text{ kg}$ y $\mu_s = 0.4$), ¿cuál es la fuerza horizontal mínima necesaria para que el bloque empiece a deslizarse?

**Solución:**
1. El bloque comenzará a deslizarse en el instante exacto en que la fuerza aplicada iguale o supere infinitesimalmente a la fuerza de fricción estática máxima.
2. $$F_{\text{mín}} = f_{s,\text{máx}} = \mu_s n$$
3. Sabiendo que $n = 98\text{ N}$:
   $$F_{\text{mín}} = 0.4 \times 98\text{ N} = 39.2\text{ N}$$
4. **Resultado:** La fuerza mínima para iniciar el movimiento es de **$39.2\text{ N}$**.

### Ejercicio 3: Bloque deslizándose a velocidad constante
Una caja de $20\text{ kg}$ es arrastrada sobre el suelo horizontal aplicando una fuerza horizontal constante de $80\text{ N}$. Si la caja se mueve con velocidad constante, ¿cuál es el coeficiente de fricción cinética ($\mu_k$) entre la caja y el suelo?

**Solución:**
1. Al moverse a **velocidad constante**, la aceleración es cero ($a = 0$). Por la Primera Ley de Newton en el eje horizontal:
   $$\Sigma F_x = F - f_k = 0 \implies f_k = F = 80\text{ N}$$
2. Determinamos la normal en el eje vertical:
   $$n = mg = (20\text{ kg})(9.8\text{ m/s}^2) = 196\text{ N}$$
3. Usamos la fórmula de la fricción cinética ($f_k = \mu_k n$) para despejar $\mu_k$:
   $$\mu_k = \frac{f_k}{n} = \frac{80\text{ N}}{196\text{ N}} \approx 0.408$$
4. **Resultado:** El coeficiente de fricción cinética es de aproximadamente **$0.41$**.

### Ejercicio 4: Desaceleración por fricción
Un disco de tejo (puck) se lanza sobre una superficie de hielo rugoso con una velocidad inicial de $12\text{ m/s}$. El coeficiente de fricción cinética es $\mu_k = 0.15$. ¿Qué distancia recorrerá el disco antes de detenerse?

**Solución:**
1. La única fuerza horizontal que actúa sobre el disco durante su viaje es la fuerza de fricción cinética ($f_k$), que se opone al movimiento:
   $$\Sigma F_x = -f_k = ma \implies -\mu_k n = ma$$
2. Como la normal es $n = mg$, sustituimos:
   $$-\mu_k mg = ma \implies a = -\mu_k g = -(0.15)(9.8\text{ m/s}^2) = -1.47\text{ m/s}^2$$
3. Usamos la ecuación de cinemática para velocidad y posición ($v_f^2 = v_0^2 + 2ad$): El disco se detiene, por lo que $v_f = 0$.
   $$0 = (12\text{ m/s})^2 + 2(-1.47\text{ m/s}^2)d$$
   $$d = \frac{-144}{-2.94} \approx 48.98\text{ m}$$
4. **Resultado:** El disco recorrerá aproximadamente **$49\text{ m}$** antes de detenerse.

### Ejercicio 5: Fuerza aplicada en un ángulo (Tirando)
Se tira de un trineo de $15\text{ kg}$ sobre nieve horizontal aplicando una fuerza de $50\text{ N}$ a un ángulo de $30^\circ$ sobre la horizontal. El coeficiente de fricción cinética es $\mu_k = 0.10$. Determine la aceleración del trineo.

**Solución:**
1. Descomponer la fuerza aplicada:
   * $F_x = F \cos(30^\circ) = 50 \cos(30^\circ) \approx 43.3\text{ N}$
   * $F_y = F \operatorname{sen}(30^\circ) = 50 \operatorname{sen}(30^\circ) = 25\text{ N}$
2. Equilibrio en el eje vertical (la componente vertical de la fuerza ayuda a levantar el trineo, reduciendo la normal):
   $$\Sigma F_y = n + F_y - mg = 0 \implies n = mg - F_y$$
   $$n = (15\text{ kg})(9.8\text{ m/s}^2) - 25\text{ N} = 147 - 25 = 122\text{ N}$$
3. Calcular la fuerza de fricción cinética:
   $$f_k = \mu_k n = (0.10)(122\text{ N}) = 12.2\text{ N}$$
4. Aplicar la Segunda Ley de Newton en el eje horizontal:
   $$\Sigma F_x = F_x - f_k = ma \implies 43.3\text{ N} - 12.2\text{ N} = (15\text{ kg})a$$
   $$a = \frac{31.1\text{ N}}{15\text{ kg}} \approx 2.07\text{ m/s}^2$$
5. **Resultado:** La aceleración del trineo es de **$2.07\text{ m/s}^2$**.

### Ejercicio 6: Fuerza aplicada en un ángulo (Empujando)
Si en lugar de tirar del trineo del Ejercicio 5 con $50\text{ N}$ a $30^\circ$, se le **empuja** con el mismo ángulo hacia abajo contra el suelo. Calcule la nueva fuerza de fricción cinética si $\mu_k = 0.10$ y el trineo tiene una masa de $15\text{ kg}$.

**Solución:**
1. Al empujar, la componente vertical de la fuerza apunta hacia abajo, aumentando la fuerza normal:
   * $F_y = 50 \operatorname{sen}(30^\circ) = 25\text{ N}$ (hacia abajo)
2. Plantear el equilibrio vertical:
   $$\Sigma F_y = n - F_y - mg = 0 \implies n = mg + F_y$$
   $$n = 147\text{ N} + 25\text{ N} = 172\text{ N}$$
3. Calcular la fuerza de fricción kinetica con la nueva normal:
   $$f_k = \mu_k n = (0.10)(172\text{ N}) = 17.2\text{ N}$$
4. **Resultado:** La nueva fuerza de fricción es **$17.2\text{ N}$** (mayor que cuando se tiraba de él).

### Ejercicio 7: Bloque en plano inclinado (Ángulo límite para deslizamiento)
Un bloque descansa sobre un plano inclinado cuyo ángulo de inclinación $\theta$ se aumenta gradualmente. El coeficiente de fricción estática es $\mu_s = 0.5$. ¿A qué ángulo crítico $\theta$ comenzará el bloque a deslizarse?

**Solución:**
1. Descomponer el peso del bloque en componentes paralelas y perpendiculares al plano:
   * $w_{\parallel} = mg \operatorname{sen}\theta$ (tiende a hacer deslizar el bloque)
   * $w_{\perp} = mg \cos\theta$ (presiona el bloque contra el plano)
2. Ecuaciones de equilibrio justo antes del deslizamiento:
   * Eje perpendicular: $n = mg \cos\theta$
   * Eje paralelo: $f_s = mg \operatorname{sen}\theta$
3. El deslizamiento inicia cuando $f_s = f_{s,\text{máx}} = \mu_s n$:
   $$mg \operatorname{sen}\theta = \mu_s (mg \cos\theta)$$
4. Dividir por $mg \cos\theta$:
   $$\frac{\operatorname{sen}\theta}{\cos\theta} = \mu_s \implies \tan\theta = \mu_s$$
5. Calcular el ángulo:
   $$\theta = \arctan(0.5) \approx 26.57^\circ$$
6. **Resultado:** El bloque comenzará a deslizarse a un ángulo aproximado de **$26.6^\circ$**.

### Ejercicio 8: Plano inclinado con movimiento ascendente
Se lanza un bloque de $5\text{ kg}$ hacia arriba por un plano inclinado a $30^\circ$ con una velocidad inicial de $8\text{ m/s}$. El coeficiente de fricción cinética es $\mu_k = 0.3$. Determine la aceleración del bloque mientras sube.

**Solución:**
1. Fuerzas que actúan sobre el bloque en dirección paralela al plano mientras sube (ambas apuntan hacia abajo del plano, oponiéndose al movimiento):
   * Componente del peso: $mg \operatorname{sen}(30^\circ)$
   * Fuerza de fricción cinética: $f_k = \mu_k n$
2. Determinar la fuerza normal:
   $$n = mg \cos(30^\circ) = (5)(9.8)\cos(30^\circ) \approx 42.44\text{ N}$$
3. Calcular la fricción:
   $$f_k = 0.3 \times 42.44\text{ N} \approx 12.73\text{ N}$$
4. Aplicar Segunda Ley de Newton paralela al plano:
   $$\Sigma F_{\parallel} = -mg \operatorname{sen}(30^\circ) - f_k = ma$$
   $$-(5)(9.8)(0.5) - 12.73 = 5a$$
   $$-24.5 - 12.73 = 5a \implies -37.23 = 5a$$
   $$a = -7.45\text{ m/s}^2$$
5. **Resultado:** La aceleración del bloque cuesta arriba es de **$-7.45\text{ m/s}^2$** (desacelerando).

### Ejercicio 9: Bloque sobre plano inclinado descendente
Si el bloque del Ejercicio 8 se detiene en su punto más alto y luego comienza a bajar por el plano de $30^\circ$. ¿Cuál es su aceleración durante el descenso? ($\mu_k = 0.3$).

**Solución:**
1. Al bajar, la componente del peso jala hacia abajo del plano, pero la fuerza de fricción cambia de sentido y ahora se opone al movimiento apuntando plano arriba:
   $$\Sigma F_{\parallel} = mg \operatorname{sen}(30^\circ) - f_k = ma$$
2. Reutilizando los valores de fuerza del ejercicio anterior:
   * Componente del peso: $mg \operatorname{sen}(30^\circ) = 24.5\text{ N}$
   * Fricción cinética: $f_k = 12.73\text{ N}$
3. Sustituyendo en la ecuación:
   $$24.5 - 12.73 = 5a$$
   $$11.77 = 5a \implies a = 2.35\text{ m/s}^2$$
4. **Resultado:** La aceleración del bloque en el descenso es de **$2.35\text{ m/s}^2$** cuesta abajo.

### Ejercicio 10: Dos bloques conectados (Máquina de Atwood con fricción)
Un bloque A de $4\text{ kg}$ está sobre una mesa horizontal áspera con $\mu_k = 0.2$. Está conectado mediante una cuerda ligera sin fricción que pasa por una polea ideal a un bloque colgante B de $2\text{ kg}$. Calcule la aceleración del sistema al liberarse.

**Solución:**
1. Fuerzas sobre el Bloque A (en la mesa):
   * Normal: $n_A = m_A g = (4\text{ kg})(9.8\text{ m/s}^2) = 39.2\text{ N}$
   * Fricción: $f_{k,A} = \mu_k n_A = (0.2)(39.2\text{ N}) = 7.84\text{ N}$
   * Segunda Ley: $T - f_{k,A} = m_A a \implies T - 7.84 = 4a$
2. Fuerzas sobre el Bloque B (colgante):
   * Segunda Ley: $m_B g - T = m_B a \implies (2)(9.8) - T = 2a \implies 19.6 - T = 2a$
3. Sumamos ambas ecuaciones para eliminar la tensión $T$:
   $$(T - 7.84) + (19.6 - T) = 4a + 2a$$
   $$11.76 = 6a \implies a = 1.96\text{ m/s}^2$$
4. **Resultado:** La aceleración del sistema es de **$1.96\text{ m/s}^2$**.

### Ejercicio 11: Dos bloques apilados
Un bloque pequeño de masa $m = 2\text{ kg}$ se coloca encima de un bloque más grande de masa $M = 8\text{ kg}$ que descansa sobre un suelo sin fricción. El coeficiente de fricción estática entre ambos bloques es $\mu_s = 0.3$. ¿Cuál es la fuerza horizontal máxima $F$ que se puede aplicar al bloque grande $M$ para que el bloque de arriba no resbale?

**Solución:**
1. Para que el bloque pequeño $m$ no resbale, debe acelerar al mismo ritmo $a$ que el bloque grande $M$.
2. La única fuerza horizontal que acelera a $m$ es la fuerza de fricción estática ejercida por el bloque $M$ sobre él:
   $$\Sigma F_x = f_s = m a$$
3. La fuerza de fricción estática máxima que puede existir entre los dos bloques es:
   $$f_{s,\text{máx}} = \mu_s n = \mu_s (mg) = (0.3)(2)(9.8) = 5.88\text{ N}$$
4. Calculamos la aceleración máxima permitida para el bloque $m$:
   $$a_{\text{máx}} = \frac{f_{s,\text{máx}}}{m} = \mu_s g = (0.3)(9.8\text{ m/s}^2) = 2.94\text{ m/s}^2$$
5. Analizando el sistema completo de masa total $(m+M)$ para hallar la fuerza máxima aplicada $F$:
   $$F = (M + m) a_{\text{máx}} = (8\text{ kg} + 2\text{ kg})(2.94\text{ m/s}^2) = 29.4\text{ N}$$
6. **Resultado:** La fuerza máxima es de **$29.4\text{ N}$**.

### Ejercicio 12: Bloque presionado contra una pared vertical
Un bloque de $3\text{ kg}$ se presiona contra una pared vertical mediante una fuerza horizontal $F$. El coeficiente de fricción estática entre la pared y el bloque es $\mu_s = 0.25$. ¿Cuál es la fuerza horizontal mínima $F$ requerida para evitar que el bloque caiga?

**Solución:**
1. Analizar fuerzas verticales para evitar el deslizamiento hacia abajo:
   $$\Sigma F_y = f_s - mg = 0 \implies f_s = mg = (3\text{ kg})(9.8\text{ m/s}^2) = 29.4\text{ N}$$
2. Analizar fuerzas horizontales: La fuerza normal $n$ de la pared es producida por la fuerza de empuje $F$:
   $$n = F$$
3. Para que no resbale, la fricción requerida debe cumplir con $f_s \le \mu_s n$:
   $$mg \le \mu_s F \implies F \ge \frac{mg}{\mu_s}$$
4. Sustituir los valores:
   $$F_{\text{mín}} = \frac{29.4\text{ N}}{0.25} = 117.6\text{ N}$$
5. **Resultado:** Se requiere aplicar una fuerza mínima de **$117.6\text{ N}$** contra la pared.

### Ejercicio 13: El coeficiente de fricción en curvas planas
Un automóvil viaja sobre una carretera horizontal plana en una curva circular de radio $R = 80\text{ m}$. ¿Cuál es el coeficiente mínimo de fricción estática necesario para que el automóvil tome la curva a una velocidad de $72\text{ km/h}$ sin derrapar?

**Solución:**
1. Convertir la velocidad a unidades del S.I.:
   $$v = 72\text{ km/h} = \frac{72}{3.6} = 20\text{ m/s}$$
2. En una curva plana, la única fuerza que actúa de manera centrípeta (hacia el centro del círculo) impidiendo el derrape lateral es la fricción estática de los neumáticos:
   $$f_s = m a_c = m \frac{v^2}{R}$$
3. Sabemos que la normal es $n = mg$. La condición de no derrapamiento es:
   $$f_s \le \mu_s n \implies m \frac{v^2}{R} \le \mu_s m g$$
4. Simplificar la masa y despejar $\mu_s$:
   $$\mu_s \ge \frac{v^2}{gR} = \frac{(20\text{ m/s})^2}{(9.8\text{ m/s}^2)(80\text{ m})} = \frac{400}{784} \approx 0.51$$
5. **Resultado:** El coeficiente mínimo de fricción estática necesario es de **$0.51$**.

### Ejercicio 14: Trabajo realizado por la fuerza de fricción
Una caja de $50\text{ kg}$ se arrastra una distancia de $10\text{ m}$ sobre un suelo horizontal rugoso aplicando una fuerza horizontal constante. Si el coeficiente de fricción cinética es $\mu_k = 0.3$, calcule el trabajo realizado por la fuerza de fricción cinética sobre la caja.

**Solución:**
1. Determinar la fuerza normal de la caja en equilibrio vertical:
   $$n = mg = (50\text{ kg})(9.8\text{ m/s}^2) = 490\text{ N}$$
2. Calcular la magnitud de la fuerza de fricción cinética:
   $$f_k = \mu_k n = (0.3)(490\text{ N}) = 147\text{ N}$$
3. Como la fricción actúa en dirección opuesta al desplazamiento ($180^\circ$), su trabajo es negativo:
   $$W_f = f_k d \cos(180^\circ) = (147\text{ N})(10\text{ m})(-1) = -1470\text{ J}$$
4. **Resultado:** El trabajo realizado por la fuerza de fricción es **$-1470\text{ Julios}$** (o $-1.47\text{ kJ}$).

### Ejercicio 15: Dos bloques y plano inclinado con poleas
Un bloque A ($m_A = 6\text{ kg}$) descansa sobre una pendiente de $37^\circ$. El coeficiente de fricción cinética entre A y la rampa es $\mu_k = 0.25$. El bloque A se conecta por un cable a un bloque B ($m_B = 8\text{ kg}$) que cuelga libremente. Determine la aceleración si el sistema se mueve y el bloque B desciende.

**Solución:**
1. Al descender el bloque B, el bloque A se mueve hacia arriba de la rampa. Por lo tanto, la fricción sobre A se dirige hacia abajo del plano.
2. Fuerzas sobre el bloque A (rampa):
   * Normal: $n_A = m_A g \cos(37^\circ) = (6)(9.8)(0.8) = 47.04\text{ N}$
   * Fricción cinética: $f_{k,A} = \mu_k n_A = (0.25)(47.04\text{ N}) = 11.76\text{ N}$
   * Componente del peso paralela a la rampa: $m_A g \operatorname{sen}(37^\circ) = (6)(9.8)(0.6) = 35.28\text{ N}$
   * Ecuación Segunda Ley: $T - f_{k,A} - m_A g \operatorname{sen}(37^\circ) = m_A a$
     $$T - 11.76 - 35.28 = 6a \implies T - 47.04 = 6a$$
3. Fuerzas sobre el bloque B (vertical colgante):
   * Ecuación Segunda Ley: $m_B g - T = m_B a$
     $$(8)(9.8) - T = 8a \implies 78.4 - T = 8a$$
4. Sumamos las dos ecuaciones para eliminar $T$:
   $$(T - 47.04) + (78.4 - T) = 6a + 8a$$
   $$31.36 = 14a \implies a = \frac{31.36}{14} = 2.24\text{ m/s}^2$$
5. **Resultado:** El bloque B desciende con una aceleración constante de **$2.24\text{ m/s}^2$**.
