# Movimiento Circular

De acuerdo con la física de Sears y Zemansky, el **movimiento circular** ocurre cuando una partícula se desplaza describiendo una trayectoria trayectoria circular de radio constante $R$.

Dependiendo de si la velocidad de la partícula cambia o no de magnitud, este movimiento se clasifica en dos tipos principales:
1. **Movimiento Circular Uniforme (MCU):** La rapidez del objeto es constante.
2. **Movimiento Circular No Uniforme (MCNU):** La rapidez del objeto varía con el tiempo.

---

## 1. Movimiento Circular Uniforme (MCU)
En el MCU, aunque la **rapidez** (la magnitud de la velocidad $v$) sea constante, el vector **velocidad** $\vec{v}$ cambia continuamente de dirección. Dado que la velocidad cambia, por definición existe una aceleración.

### A. Aceleración Centrípeta ($a_{rad}$ o $a_c$)
Esta aceleración es siempre perpendicular a la velocidad de la partícula y apunta radialmente hacia el centro del círculo. Su función es cambiar únicamente la dirección del vector velocidad, no su magnitud. Su fórmula es:

$$a_{rad} = \frac{v^2}{R}$$

También podemos expresarla en términos del **período** ($T$), que es el tiempo que tarda el objeto en dar una vuelta completa:

$$v = \frac{2\pi R}{T} \implies a_{rad} = \frac{4\pi^2 R}{T^2}$$

### B. Dinámica del Movimiento Circular Uniforme
Para que un cuerpo de masa $m$ tenga una aceleración centrípeta, la Segunda Ley de Newton ($F_{net} = ma$) nos dice que debe existir una fuerza neta dirigida hacia el centro de la trayectoria, denominada **fuerza centrípeta** o **fuerza radial neta** ($F_{rad}$):

$$\sum F_{rad} = m a_{rad} = m\frac{v^2}{R}$$

*Nota de Sears-Zemansky:* La "fuerza centrípeta" **no es un nuevo tipo de fuerza** (como la gravedad o la tensión). Es el rol que juegan las fuerzas reales ya existentes (tensión de una cuerda, fricción, gravedad, normal, etc.) para mantener al objeto en su trayectoria circular.

---

## 2. Movimiento Circular No Uniforme
Si la rapidez del cuerpo también cambia con el tiempo, el vector aceleración $\vec{a}$ ya no apunta directamente al centro de la trayectoria. En este caso, la aceleración total se descompone en dos componentes ortogonales:

1. **Aceleración Radial (o Centrípeta) $a_{rad}$:** Mide el cambio en la *dirección* de la velocidad.
   $$a_{rad} = \frac{v^2}{R}$$
2. **Aceleración Tangencial $a_{tan}$:** Mide el cambio en la *magnitud* de la velocidad (la rapidez). Es colineal con el vector velocidad:
   $$a_{tan} = \frac{dv}{dt}$$

### Aceleración Total
Al ser componentes mutuamente perpendiculares, la magnitud de la aceleración total del cuerpo es:

$$a = \sqrt{a_{rad}^2 + a_{tan}^2}$$

---

## 3. Variables Angulares (Relación lineal-angular)
Para describir de manera más sencilla el movimiento circular, a menudo se utilizan variables angulares:

* **Posición angular ($\theta$):** Medida en radianes ($rad$).
* **Velocidad angular ($\omega$):** Tasa de cambio del ángulo respecto al tiempo ($\omega = \frac{d\theta}{dt}$), expresada en $rad/s$.
* **Aceleración angular ($\alpha$):** Tasa de cambio de la velocidad angular ($\alpha = \frac{d\omega}{dt}$), expresada en $rad/s^2$.

### Fórmulas de Conexión:
* Velocidad lineal: $v = \omega R$
* Aceleración tangencial: $a_{tan} = \alpha R$
* Aceleración radial: $a_{rad} = \omega^2 R$

### Ejercicio 1: Conversión de Variables Angulares y Velocidad Lineal
**Enunciado:** Un volante de $0.400\text{ m}$ de diámetro gira con una velocidad angular constante de $120\text{ rpm}$ (revoluciones por minuto).
1. Calcule su velocidad angular en $\text{rad/s}$.
2. Halle la rapidez lineal de un punto en el borde del volante.

**Solución:**
1. **Velocidad angular ($\omega$):**
   Convertimos las revoluciones por minuto a radianes por segundo:
   $$\omega = 120\text{ rpm} = 120 \cdot \frac{2\pi\text{ rad}}{60\text{ s}} = 4\pi\text{ rad/s} \approx 12.57\text{ rad/s}$$

2. **Rapidez lineal ($v$):**
   El radio es la mitad del diámetro, $R = \frac{0.400\text{ m}}{2} = 0.200\text{ m}$. Usamos la relación $v = \omega R$:
   $$v = (12.57\text{ rad/s})(0.200\text{ m}) = 2.51\text{ m/s}$$

### Ejercicio 2: Aceleración Radial en la Tierra
**Enunciado:** Calcule la aceleración radial (centrípeta) de un objeto situado en el ecuador de la Tierra debido a la rotación del planeta sobre su propio eje. Considere el radio ecuatorial terrestre como $R = 6.38 \times 10^6\text{ m}$ y el período de rotación de 24 horas.

**Solución:**
1. **Periodo en segundos ($T$):**
   $$T = 24\text{ h} = 24 \times 3600\text{ s} = 86,400\text{ s}$$

2. **Aceleración radial ($a_{rad}$):**
   Usamos la relación en función del período:
   $$a_{rad} = \frac{4\pi^2 R}{T^2} = \frac{4\pi^2 (6.38 \times 10^6\text{ m})}{(86,400\text{ s})^2}$$
   $$a_{rad} = \frac{2.518 \times 10^8}{7.465 \times 10^9} \approx 0.0337\text{ m/s}^2$$
   *Nota:* Este valor es aproximadamente el $0.34\%$ de la aceleración de la gravedad ($g$).

### Ejercicio 3: Fuerza Centrípeta en un Satélite Orbitando
**Enunciado:** Un satélite de $500\text{ kg}$ se encuentra en una órbita circular alrededor de la Tierra a una altitud de $1.60 \times 10^6\text{ m}$ sobre la superficie terrestre. El radio de la Tierra es $6.38 \times 10^6\text{ m}$ y el campo gravitacional a esa altitud es $g' = 6.20\text{ m/s}^2$. Calcule la rapidez del satélite.

**Solución:**
1. **Radio de la órbita ($R$):**
   Es la suma del radio terrestre y la altitud:
   $$R = R_T + h = 6.38 \times 10^6\text{ m} + 1.60 \times 10^6\text{ m} = 7.98 \times 10^6\text{ m}$$

2. **Fuerza centrípeta:**
   La única fuerza que actúa radialmente en la órbita es la fuerza de gravedad $F_g = m g'$:
   $$\sum F_{rad} = m g' = m \frac{v^2}{R}$$
   Podemos simplificar la masa $m$:
   $$g' = \frac{v^2}{R} \implies v = \sqrt{g' R}$$

3. **Cálculo final:**
   $$v = \sqrt{(6.20\text{ m/s}^2)(7.98 \times 10^6\text{ m})} = \sqrt{4.948 \times 10^7} \approx 7,034\text{ m/s} = 7.03\text{ km/s}$$

### Ejercicio 4: Curva Plana sin Peralte
**Enunciado:** Un automóvil viaja por una curva horizontal plana de radio $R = 150\text{ m}$. Si el coeficiente de fricción estática entre los neumáticos y el asfalto seco es $\mu_s = 0.80$, ¿cuál es la rapidez máxima con la que el auto puede tomar la curva sin derrapar?

**Solución:**
1. **Fuerzas en juego:**
   * En el eje vertical: $n - mg = 0 \implies n = mg$.
   * En el eje radial horizontal: la fricción estática máxima suministra la aceleración centrípeta necesaria:
     $$\sum F_{rad} = f_{s,\text{máx}} = m \frac{v_{\text{máx}}^2}{R}$$

2. **Uso de la ecuación de fricción:**
   Sabemos que $f_{s,\text{máx}} = \mu_s n = \mu_s mg$. Sustituyendo:
   $$\mu_s mg = m \frac{v_{\text{máx}}^2}{R} \implies \mu_s g = \frac{v_{\text{máx}}^2}{R}$$
   $$v_{\text{máx}} = \sqrt{\mu_s g R}$$

3. **Cálculo numérico:**
   $$v_{\text{máx}} = \sqrt{(0.80)(9.80\text{ m/s}^2)(150\text{ m})} = \sqrt{1176} \approx 34.3\text{ m/s}\text{ (aprox. } 123\text{ km/h)}$$

### Ejercicio 5: Curva con Peralte (Diseño de Carreteras)
**Enunciado:** Una curva de autopista de radio $250\text{ m}$ está diseñada para una rapidez de $25\text{ m/s}$. ¿Qué ángulo de peralte debe tener para que los vehículos no dependan de la fricción para mantenerse en la curva?

**Solución:**
1. **Condición de peralte ideal:**
   Al no haber fricción, la componente horizontal de la fuerza normal proporciona la aceleración centrípeta:
   $$n \sin\theta = m\frac{v^2}{R}$$
   La componente vertical de la normal equilibra el peso del auto:
   $$n \cos\theta = mg$$

2. **Relación trigonométrica:**
   Dividimos la primera ecuación entre la segunda:
   $$\frac{n \sin\theta}{n \cos\theta} = \frac{m v^2 / R}{mg} \implies \tan\theta = \frac{v^2}{g R}$$

3. **Cálculo del ángulo $\theta$:**
   $$\tan\theta = \frac{(25\text{ m/s})^2}{(9.80\text{ m/s}^2)(250\text{ m})} = \frac{625}{2450} \approx 0.2551$$
   $$\theta = \arctan(0.2551) \approx 14.3^\circ$$

### Ejercicio 6: El Péndulo Cónico
**Enunciado:** Una pequeña esfera de masa $m$ está suspendida de una cuerda ligera de longitud $L = 1.20\text{ m}$ y describe un círculo horizontal con rapidez constante. La cuerda forma un ángulo constante de $\beta = 30.0^\circ$ con la vertical. Calcule la rapidez lineal de la esfera.

**Solución:**
1. **Radio de la trayectoria ($R$):**
   Por trigonometría, $R = L \sin\beta = (1.20\text{ m})\sin(30.0^\circ) = 0.60\text{ m}$.

2. **Diagrama de Cuerpo Libre (DCL):**
   * Eje vertical: $T \cos\beta - mg = 0 \implies T \cos\beta = mg$.
   * Eje radial horizontal: la componente de la tensión tira hacia el centro:
     $$T \sin\beta = m\frac{v^2}{R}$$

3. **Combinación de ecuaciones:**
   Dividiendo las ecuaciones:
   $$\tan\beta = \frac{v^2}{g R} \implies v = \sqrt{g R \tan\beta}$$

4. **Cálculo numérico:**
   $$v = \sqrt{(9.80\text{ m/s}^2)(0.60\text{ m})\tan(30.0^\circ)} = \sqrt{5.88 \times 0.5774} = \sqrt{3.395} \approx 1.84\text{ m/s}$$

### Ejercicio 7: El Cubo de Agua Giratorio en un Plano Vertical
**Enunciado:** Un acróbata hace girar un balde de agua en un círculo vertical de radio $R = 0.90\text{ m}$. ¿Cuál es la rapidez mínima que debe tener el balde en el punto más alto del trayecto para que el agua no se derrame?

**Solución:**
1. **Análisis en el punto más alto:**
   En la parte superior, tanto el peso del agua ($mg$) como la fuerza normal del fondo del balde ($n$) apuntan hacia abajo (dirección radial hacia el centro):
   $$\sum F_{rad} = n + mg = m\frac{v^2}{R}$$

2. **Condición de rapidez crítica (mínima):**
   Para que el agua apenas permanezca en contacto con el fondo, la fuerza normal tiende a cero ($n \approx 0$):
   $$mg = m\frac{v_{\text{mín}}^2}{R} \implies v_{\text{mín}} = \sqrt{g R}$$

3. **Cálculo numérico:**
   $$v_{\text{mín}} = \sqrt{(9.80\text{ m/s}^2)(0.90\text{ m})} = \sqrt{8.82} \approx 2.97\text{ m/s}$$

### Ejercicio 8: Rueda de la Fortuna (Movimiento Vertical)
**Enunciado:** Un pasajero de $60.0\text{ kg}$ viaja en una rueda de la fortuna de $15.0\text{ m}$ de radio que gira uniformemente a una rapidez de $3.00\text{ m/s}$. Halle la fuerza normal que ejerce el asiento sobre el pasajero en:
1. El punto más bajo de la trayectoria.
2. El punto más alto de la trayectoria.

**Solución:**
1. **Punto más bajo:**
   La fuerza normal $n_1$ es hacia arriba y el peso es hacia abajo. La aceleración centrípeta apunta hacia arriba (al centro):
   $$\sum F_{rad} = n_1 - mg = m\frac{v^2}{R} \implies n_1 = m\left(g + \frac{v^2}{R}\right)$$
   $$n_1 = 60.0\left(9.80 + \frac{3.00^2}{15.0}\right) = 60.0(9.80 + 0.60) = 60.0(10.4) = 624\text{ N}$$

2. **Punto más alto:**
   La normal $n_2$ es hacia arriba, pero la aceleración centrípeta apunta hacia abajo (hacia el centro):
   $$\sum F_{rad} = mg - n_2 = m\frac{v^2}{R} \implies n_2 = m\left(g - \frac{v^2}{R}\right)$$
   $$n_2 = 60.0(9.80 - 0.60) = 60.0(9.20) = 552\text{ N}$$

### Ejercicio 9: Movimiento Circular No Uniforme
**Enunciado:** Una partícula se desplaza en una trayectoria circular de radio $R = 2.00\text{ m}$. En un instante dado, su rapidez es de $4.00\text{ m/s}$ y su rapidez aumenta a una tasa de $3.00\text{ m/s}^2$. Halle las componentes de la aceleración y la magnitud de la aceleración total.

**Solución:**
1. **Aceleración radial (centrípeta):**
   $$a_{rad} = \frac{v^2}{R} = \frac{(4.00\text{ m/s})^2}{2.00\text{ m}} = 8.00\text{ m/s}^2$$

2. **Aceleración tangencial:**
   La tasa de incremento de la rapidez es directamente la aceleración tangencial:
   $$a_{tan} = 3.00\text{ m/s}^2$$

3. **Magnitud de la aceleración total ($a$):**
   $$a = \sqrt{a_{rad}^2 + a_{tan}^2} = \sqrt{8.00^2 + 3.00^2} = \sqrt{64.00 + 9.00} = \sqrt{73.00} \approx 8.54\text{ m/s}^2$$

### Ejercicio 10: Ultracentrífuga de Laboratorio
**Enunciado:** Una ultracentrífuga gira a $60,000\text{ rpm}$. Una muestra de prueba se encuentra a $8.00\text{ cm}$ del eje de rotación. Calcule la aceleración centrípeta en términos de "g" (donde $g = 9.80\text{ m/s}^2$).

**Solución:**
1. **Conversión de la velocidad angular ($\omega$):**
   $$\omega = 60,000\text{ rpm} = 60,000 \times \frac{2\pi\text{ rad}}{60\text{ s}} = 2000\pi\text{ rad/s} \approx 6,283.18\text{ rad/s}$$

2. **Radio ($R$):**
   $$R = 8.00\text{ cm} = 0.0800\text{ m}$$

3. **Aceleración radial ($a_{rad}$):**
   $$a_{rad} = \omega^2 R = (6,283.18\text{ rad/s})^2 (0.0800\text{ m}) \approx 3.1583 \times 10^6\text{ m/s}^2$$

4. **Expresión en términos de $g$:**
   $$\text{Múltiplo de } g = \frac{a_{rad}}{g} = \frac{3.1583 \times 10^6\text{ m/s}^2}{9.80\text{ m/s}^2} \approx 322,000 g$$

### Ejercicio 11: El Rotor de Parque de Diversiones
**Enunciado:** En un parque de atracciones se monta un gran cilindro ("Rotor") de radio $R = 3.00\text{ m}$. Las personas se apoyan de espalda en la pared cilíndrica. El cilindro comienza a girar rápidamente y el piso desciende por completo. Si el coeficiente de fricción estática entre la ropa y la pared es $\mu_s = 0.400$, ¿cuál es el periodo mínimo de rotación para que las personas no caigan?

**Solución:**
1. **Diagrama de Fuerzas:**
   * Eje vertical: la fricción estática sostiene el peso de la persona: $f_s - mg = 0 \implies f_s = mg$.
   * Eje radial horizontal: la pared empuja hacia adentro proporcionando la fuerza centrípeta: $n = m\omega^2 R$.

2. **Condición límite de deslizamiento:**
   $$f_s \le \mu_s n \implies mg \le \mu_s (m \omega^2 R)$$
   Cancelando la masa $m$:
   $$g \le \mu_s \omega^2 R \implies \omega^2 \ge \frac{g}{\mu_s R}$$
   La velocidad angular mínima requerida es:
   $$\omega_{\text{mín}} = \sqrt{\frac{g}{\mu_s R}}$$

3. **Determinando el periodo máximo de rotación ($T$):**
   Sabiendo que $\omega = 2\pi / T$:
   $$\frac{2\pi}{T_{\text{máx}}} = \sqrt{\frac{g}{\mu_s R}} \implies T_{\text{máx}} = 2\pi \sqrt{\frac{\mu_s R}{g}}$$

4. **Cálculo numérico:**
   $$T_{\text{máx}} = 2\pi \sqrt{\frac{(0.400)(3.00\text{ m})}{9.80\text{ m/s}^2}} = 2\pi \sqrt{0.1224} \approx 2\pi (0.350) \approx 2.20\text{ s}$$

### Ejercicio 12: Cinemática Angular con Aceleración Constante
**Enunciado:** La hoja de una sierra circular de $0.20\text{ m}$ de radio arranca desde el reposo y experimenta una aceleración angular constante de $\alpha = 4.0\text{ rad/s}^2$. Para $t = 3.0\text{ s}$, determine:
1. La rapidez angular final.
2. El ángulo total girado en radianes.

**Solución:**
1. **Rapidez angular final ($\omega$):**
   $$\omega = \omega_0 + \alpha t = 0 + (4.0\text{ rad/s}^2)(3.0\text{ s}) = 12\text{ rad/s}$$

2. **Ángulo barrido ($\theta$):**
   $$\theta = \theta_0 + \omega_0 t + \frac{1}{2}\alpha t^2$$
   $$\theta = 0 + 0 + \frac{1}{2}(4.0\text{ rad/s}^2)(3.0\text{ s})^2 = 2.0 \times 9.0 = 18\text{ rad}$$

### Ejercicio 13: Cuerda en una Mesa sin Fricción
**Enunciado:** Un bloque de masa $m_1 = 0.500\text{ kg}$ sobre una mesa horizontal sin fricción está conectado por una cuerda que pasa por un orificio central a un objeto colgante de masa $m_2 = 1.50\text{ kg}$. ¿Con qué rapidez constante $v$ debe girar $m_1$ en un círculo de radio $R = 0.300\text{ m}$ para que la masa colgante $m_2$ permanezca en reposo?

**Solución:**
1. **Análisis de la masa colgante $m_2$:**
   Para que no se mueva verticalmente, la tensión de la cuerda $T$ debe equilibrar su peso:
   $$T = m_2 g$$

2. **Análisis del bloque en rotación $m_1$:**
   La tensión de la cuerda suministra en su totalidad la fuerza radial necesaria:
   $$\sum F_{rad} = T = m_1 \frac{v^2}{R}$$

3. **Unificación de ecuaciones:**
   Sustituimos la tensión:
   $$m_2 g = m_1 \frac{v^2}{R} \implies v = \sqrt{\frac{m_2 g R}{m_1}}$$

4. **Cálculo:**
   $$v = \sqrt{\frac{(1.50\text{ kg})(9.80\text{ m/s}^2)(0.300\text{ m})}{0.500\text{ kg}}} = \sqrt{3 \times 2.94} = \sqrt{8.82} \approx 2.97\text{ m/s}$$

### Ejercicio 14: Fuerza de fricción necesaria para girar en bicicleta
**Enunciado:** Un ciclista viaja en una pista circular de $40.0\text{ m}$ de radio a una rapidez constante de $12.0\text{ m/s}$. El ciclista y su bicicleta tienen una masa combinada de $85.0\text{ kg}$. ¿Cuál debe ser la magnitud mínima de la fuerza de fricción estática total entre las llantas y el suelo para evitar que la bicicleta deslice lateralmente?

**Solución:**
1. **Fuerzas radiales:**
   La única fuerza que empuja radialmente en una superficie plana es la fuerza de fricción estática lateral $f_s$:
   $$f_s = m \frac{v^2}{R}$$

2. **Sustitución de datos directos:**
   $$f_s = (85.0\text{ kg}) \frac{(12.0\text{ m/s})^2}{40.0\text{ m}}$$
   $$f_s = 85.0 \times \frac{144.0}{40.0} = 85.0 \times 3.60 = 306\text{ N}$$
   *La magnitud de la fuerza de fricción estática necesaria es de $306\text{ N}$.*

### Ejercicio 15: Loop-the-Loop en una Montaña Rusa
**Enunciado:** Un carro en una montaña rusa viaja sin motor a través de un bucle circular de radio $R = 12.0\text{ m}$ en un plano vertical. Si el carro tiene una rapidez de $18.0\text{ m/s}$ al pasar por la parte inferior del bucle, ¿qué aceleración experimentarán los pasajeros en esa parte?

**Solución:**
1. **Aceleración en el punto más bajo:**
   La velocidad en ese punto es tangencial y la rapidez instantánea es de $18.0\text{ m/s}$. Al no haber fuerzas propulsoras, asumimos que no hay aceleración tangencial relevante instantáneamente, o únicamente calculamos la componente radial.

2. **Cálculo de la aceleración radial (centrípeta):**
   $$a_{rad} = \frac{v^2}{R} = \frac{(18.0\text{ m/s})^2}{12.0\text{ m}} = \frac{324}{12.0} = 27.0\text{ m/s}^2$$

3. **En términos de fuerza G:**
   Podemos cuantificar la sensación física comparándola con la gravedad regular:
   $$\text{G-force} = \frac{a_{rad}}{g} = \frac{27.0\text{ m/s}^2}{9.80\text{ m/s}^2} \approx 2.76 g$$
   *(Esto representa una aceleración muy normal y emocionante para un juego mecánico comercial).*
